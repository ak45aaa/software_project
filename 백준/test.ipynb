```python
n = 3
k = 1
low = (n-k)//2
high = (n+k)//2

def split_sqr(sqr):
    temp_list = []
    rt_sqr = []
    rt_row = []
    for row in sqr:
        for x in row:
            temp = [[x for _ in range(n)] for _ in range(3)]
            for i in range(low, high):
                for j in range(low, high):
                    temp[i][j] = 1
            temp_list.append(temp)
            
        for i in range(n):
            for j in range(n):
                rt_row.append(temp_list[j][i])
            rt_sqr.append(rt_row)
            rt_row = []
        
        temp_list = []
    
    return rt_sqr
```

```python
t = int(input())

for _ in range(t):
    vps = input()
    flag = True

    tmp = 0

    for s in vps:
        if s == '(':
            tmp += 1
        else:
            tmp -= 1
        if tmp < 0:
            print("NO")
            flag = False
            break
            
    if flag:
        if tmp == 0:
            print("YES")
        else:
            print("NO")
```
